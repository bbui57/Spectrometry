## Universal Functions (UFuncs)

```python
def compute_reciprocals(values):
    output = np.empty(len(values))
    for i in range(len(values)):
        output[i] = 1.0 / values[i]
    return output
```

```python
big_array = np.random.randint(1,100,size=1000000)
print("This is a slow function:")
%timeit compute_reciprocals(big_array)
print("\nThis is the faster ufunc:")
%timeit (1.0 / big_array)
%timeit np.reciprocal(big_array.astype(float))
```

## Aggregations

### Sum

```python
print("Sum all of the values in an array:")
L = np.random.random(100)
print(L[:5])
print(sum(L))
print("\nIt also has a UFunc:")
%timeit sum(L)
%timeit np.sum(L)
```

### Minimum and Maximum

```python
print("Python's built-in functions: ", min(big_array), max(big_array))
%timeit min(big_array)
print("NumPy's UFuncs: ", np.min(big_array), np.max(big_array))
%timeit np.min(big_array)
```

```python
print("A shorter syntax is to use methods of the array object itself:")
print(big_array.min(), big_array.max(), big_array.sum())
%timeit big_array.min()        
```

### Multi-Dimensional Aggregates

```python
M = np.random.random((3,4))
print("Random multi-dimension array:\n", M)
print("Sum: ", M.sum())
print("Minimum of axis = 0 (each column): ", M.min(axis=0))
print("Maximum of axis = 1 (each row): ", M.max(axis=1))
```

### Example Application

```python
data = pd.read_csv("president_heights.csv")
data.head()
```

```python
heights = np.array(data['height(cm)'])
print(heights)
```

```python
print("Mean Height:    ", heights.mean())
print("Std Deviation:  ", heights.std())
print("Minimum Height: ", heights.min())
print("Maximum Height: ", heights.max())
```

```python
print("25th Percentile: ", np.percentile(heights, 25))
print("Median:          ", np.median(heights))
print("75th Percentile: ", np.percentile(heights, 75))
```
